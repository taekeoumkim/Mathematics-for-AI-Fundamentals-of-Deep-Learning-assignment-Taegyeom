### 수식을 읽는 방법

- `ŷ=wx+b`는 **w와 x를 곱한 뒤 b를 더한다**는 뜻입니다. 붙여 쓴 wx는 곱셈입니다.
- `L=(y−ŷ)²`는 **정답에서 예측을 뺀 값을 자기 자신과 곱한다**는 뜻입니다. 예를 들어 (10−7)²=3×3=9입니다.
- `∂L/∂w`는 w에 대한 손실의 기울기입니다. 기호가 분수처럼 생겼다고 해서 L을 w로 나누라는 뜻은 아닙니다. 코드에서는 `grad_w`라고 씁니다.
- `w_new=w_old−α×grad_w`는 **현재 w에서 학습률과 기울기의 곱을 뺀다**는 뜻입니다.
- `3−0.05×(−12)=3−(−0.6)=3+0.6=3.6`입니다. 음수를 빼면 더하기가 됩니다.
- `≈`는 ‘대략 같다’는 뜻입니다. 표의 긴 소수는 보기 좋게 반올림한 값일 수 있습니다.

학습률이 0.05라고 해서 언제나 0.05만큼 이동하지는 않습니다. 실제 변화량은 **−학습률×현재 기울기**입니다. 기울기를 계산하는 과정과 파라미터를 실제로 바꾸는 과정도 구별하세요.

## 필수 실습 1: 업데이트와 학습률

### 문제 1-1. 가중치와 편향을 두 번 업데이트하기

### 문제 설명

모델은 ŷ=wx+b입니다. x=2, y=10, 처음 w=3, b=1, 학습률 α=0.05를 사용합니다. 업데이트는 정확히 두 번만 합니다.

### 제공 공식

| 순서 | 계산 |
|---|---|
| 1 | 예측 ŷ=wx+b |
| 2 | 손실 L=(y−ŷ)² |
| 3 | w의 기울기 g_w=2(ŷ−y)x, b의 기울기 g_b=2(ŷ−y) |
| 4 | 새 w=w−0.05×g_w, 새 b=b−0.05×g_b |
| 5 | 새 w와 b로 예측과 손실을 다시 계산 |

**두 기울기는 같은 업데이트 전 상태에서 모두 계산합니다.** 그다음 두 값을 바꿉니다. 두 번째 업데이트에서는 새 상태의 기울기를 다시 구하세요.

### 요구사항

- 초기 예측·손실·두 기울기를 계산합니다.
- 1회와 2회 업데이트 후 w, b, 예측, 손실을 계산합니다.
- 공식에 대입한 값을 적고 Q1~Q4에 답합니다.

#### 해석 질문

**Q1.** 업데이트 전 예측, 손실, 두 기울기는 얼마인가요?

**Q2.** 첫 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**Q3.** 두 번째 업데이트에서 사용할 기울기와 업데이트 후 결과는 무엇인가요?

**Q4.** 기울기는 언제 구하고, 업데이트 후 손실은 언제 구하나요?

#### 제출 결과

- 요구사항의 계산 결과 또는 제공 코드의 실행 결과
- Q1~Q4 답변

```python
# 문제 1-1. 두 기울기를 같은 상태에서 구한 뒤 함께 갱신합니다.
x, y = 2.0, 10.0
w, b = 3.0, 1.0
learning_rate = 0.05
initial_prediction = w * x + b
initial_loss = (y - initial_prediction) ** 2
initial_grad_w = 2 * (initial_prediction - y) * x
initial_grad_b = 2 * (initial_prediction - y)

w1 = w - learning_rate * initial_grad_w
b1 = b - learning_rate * initial_grad_b
prediction1 = w1 * x + b1
loss1 = (y - prediction1) ** 2

# 두 번째에는 첫 갱신 후 상태에서 기울기를 다시 계산합니다.
grad_w1 = 2 * (prediction1 - y) * x
grad_b1 = 2 * (prediction1 - y)
w2 = w1 - learning_rate * grad_w1
b2 = b1 - learning_rate * grad_b1
prediction2 = w2 * x + b2
loss2 = (y - prediction2) ** 2

print("초기 예측, 손실:", initial_prediction, initial_loss)
print("초기 기울기 [w, b]:", np.round([initial_grad_w, initial_grad_b], 6))
print("1회 후 [w, b, 예측, 손실]:", np.round([w1, b1, prediction1, loss1], 6))
print("두 번째에 사용할 기울기:", np.round([grad_w1, grad_b1], 6))
print("2회 후 [w, b, 예측, 손실]:", np.round([w2, b2, prediction2, loss2], 6))
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 초기: $\hat y=3\times2+1=7$, $L=(10-7)^2=9$, $g_w=2(7-10)\times2=-12$, $g_b=2(7-10)=-6$입니다.
- 1회 후: $w=3-0.05(-12)=3.6$, $b=1-0.05(-6)=1.3$입니다. 새 예측은 8.5, 새 손실은 2.25입니다.
- 두 번째 기울기: $g_w=2(8.5-10)\times2=-6$, $g_b=2(8.5-10)=-3$입니다.
- 2회 후: $w=3.6-0.05(-6)=3.9$, $b=1.3-0.05(-3)=1.45$입니다. 새 예측은 $3.9\times2+1.45=9.25$, 손실은 $(10-9.25)^2=0.5625$입니다.

**Q1.** 업데이트 전 예측, 손실, 두 기울기는 얼마인가요?

**답변:** 초기 예측은 7, 손실은 9, [w,b] 기울기는 `[-12, -6]`입니다.

**Q2.** 첫 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**답변:** w=3.6, b=1.3, 예측=8.5, 손실=2.25입니다.

**Q3.** 두 번째 업데이트에서 사용할 기울기와 업데이트 후 결과는 무엇인가요?

**답변:** 첫 갱신 후 상태에서 구한 기울기는 `[-6, -3]`입니다. 이를 사용하면 w=3.9, b=1.45, 예측=9.25, 손실=0.5625입니다.

**Q4.** 기울기는 언제 구하고, 업데이트 후 손실은 언제 구하나요?

**답변:** 두 기울기는 모두 같은 업데이트 전 파라미터와 예측값에서 구합니다. 그 기울기로 w와 b를 갱신한 다음, 새 파라미터로 예측과 손실을 다시 계산합니다.

### 문제 1-2. 학습률에 따른 세 번의 이동 비교하기

### 문제 설명

이번에는 편향 b 없이 ŷ=2w를 사용합니다. 손실은 L=(10−2w)²이고 기울기는 g=8w−40입니다. 각각 처음 w=3에서 시작하여 학습률 0.05, 0.2, 0.25, 0.3으로 세 번씩 업데이트합니다.

### 제공 공식

- 새 w = 현재 w − 학습률×(8×현재 w−40)
- 새 손실 = (10−2×새 w)²
- w=5이면 예측이 10이 되어 손실이 0입니다.

### 요구사항

- 아래 제공 코드를 그대로 실행합니다. 반복 코드를 새로 만들 필요는 없습니다.
- 출력의 0회는 초기 상태, 1~3회는 해당 횟수만큼 바꾼 뒤의 상태입니다.
- 학습률 0.05의 첫 업데이트만 손으로 확인하고, 나머지는 출력 표를 읽어 답합니다.
- 수렴 조건식을 유도하거나 외우지 않아도 됩니다.

#### 해석 질문

**Q1.** 처음 기울기와 학습률 0.05의 첫 업데이트 후 w, 손실은 얼마인가요?

**Q2.** 학습률 0.05와 0.2에서 w의 움직임과 손실은 어떻게 다른가요?

**Q3.** 학습률 0.25에서는 가중치와 손실이 어떻게 변하나요?

**Q4.** 학습률 0.3의 결과에서 무엇을 알 수 있나요?

#### 제출 결과

- 요구사항의 계산 결과 또는 제공 코드의 실행 결과
- Q1~Q4 답변

```python
learning_rates = [0.05, 0.2, 0.25, 0.3]
for learning_rate in learning_rates:
    w = 3.0  # 학습률마다 같은 초기값으로 다시 시작
    print("학습률:", learning_rate)
    print("0회", w, (10-2*w)**2)
    for step in range(1, 4):
        grad = 8*w-40
        w = w-learning_rate*grad
        loss = (10-2*w)**2
        print(step, "회 / w, 손실:", np.round([w, loss], 6))
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 초기 w=3에서 $g=8\times3-40=-16$, 손실은 16입니다. 학습률 0.05의 첫 갱신은 $w=3-0.05(-16)=3.8$, 손실은 $(10-7.6)^2=5.76$입니다.

| 학습률 | w: 초기 → 1회 → 2회 → 3회 | 손실: 초기 → 1회 → 2회 → 3회 |
|---|---|---|
| 0.05 | 3 → 3.8 → 4.28 → 4.568 | 16 → 5.76 → 2.0736 → 0.746496 |
| 0.2 | 3 → 6.2 → 4.28 → 5.432 | 16 → 5.76 → 2.0736 → 0.746496 |
| 0.25 | 3 → 7 → 3 → 7 | 16 → 16 → 16 → 16 |
| 0.3 | 3 → 7.8 → 1.08 → 10.488 | 16 → 31.36 → 61.4656 → 120.472576 |

**Q1.** 처음 기울기와 학습률 0.05의 첫 업데이트 후 w, 손실은 얼마인가요?

**답변:** 처음 기울기는 −16입니다. 학습률 0.05의 첫 갱신 후 w는 3.8, 손실은 5.76입니다.

**Q2.** 학습률 0.05와 0.2에서 w의 움직임과 손실은 어떻게 다른가요?

**답변:** 0.05에서는 w가 5보다 작은 쪽에서 단조롭게 5에 가까워집니다. 0.2에서는 5를 번갈아 넘으며 가까워집니다. 다만 이 예제의 세 번 업데이트 후 손실은 두 학습률 모두 5.76, 2.0736, 0.746496으로 같습니다.

**Q3.** 학습률 0.25에서는 가중치와 손실이 어떻게 변하나요?

**답변:** w는 3과 7 사이를 반복하고 손실은 계속 16입니다. 최솟값을 지나쳐 같은 폭으로 왕복하므로 수렴하지 않습니다.

**Q4.** 학습률 0.3의 결과에서 무엇을 알 수 있나요?

**답변:** w가 7.8, 1.08, 10.488로 움직이며 5에서 멀어지고, 손실도 31.36, 61.4656, 120.472576으로 커집니다. 이 손실과 초기값에서는 학습률 0.3이 너무 커 발산하는 움직임입니다.

## 필수 실습 2: 데이터 묶음과 수렴

### 문제 2-1. 전체 데이터와 작은 묶음으로 업데이트하기

### 문제 설명

입력 X=[1,2,3,4], 정답 y=[2,4,6,8], 모델 ŷ=wX를 사용합니다. 처음 w=0, 학습률 0.01입니다. 편향 b는 없습니다.

- Batch: 네 데이터를 모두 사용해 한 번 업데이트합니다.
- Mini-Batch: [1,2] 입력의 묶음, [3,4] 입력의 묶음 순서로 사용해 두 번 업데이트합니다. 각 입력의 정답은 각각 [2,4], [6,8]입니다.
- 두 방식을 비교할 때 각각 w=0에서 다시 시작합니다. 데이터 순서는 섞지 않습니다.

### 제공 공식

데이터 하나의 기울기 = 2×(w×입력−정답)×입력

묶음의 기울기 = 그 묶음에 속한 각 기울기의 평균

새 w = 현재 w−0.01×묶음의 기울기

MSE = 각 데이터의 (정답−예측)²을 구한 뒤 평균

### 요구사항

1. 초기 w=0에서 네 데이터의 기울기를 구합니다.
2. Batch의 평균 기울기, 새 w, 업데이트 후 전체 MSE를 구합니다.
3. Mini-Batch의 첫 묶음으로 업데이트한 뒤, **바뀐 w**로 두 번째 묶음의 기울기를 구해 다시 업데이트합니다. 마지막 전체 MSE는 필수 계산이 아닙니다.
4. 한 Epoch의 업데이트 횟수를 비교합니다. SGD는 데이터 하나마다 업데이트한다는 뜻만 사용하며 전체 SGD 계산 코드는 작성하지 않아도 됩니다.

배열 계산을 사용하면 `grad_each = 2*(w*X-y)*X`, 평균은 `np.mean(grad_each)`입니다. 평균은 합계를 데이터 개수로 나눈 값입니다.

#### 해석 질문

**Q1.** 초기 w=0에서 네 데이터의 기울기와 전체 평균은 얼마인가요?

**Q2.** Batch로 한 번 업데이트한 w와 전체 MSE는 얼마인가요?

**Q3.** Mini-Batch의 첫 묶음과 두 번째 묶음에서 평균 기울기와 새 w는 얼마인가요?

**Q4.** 한 Epoch 동안 Batch, SGD, 크기 2의 Mini-Batch는 각각 몇 번 업데이트하나요?

#### 제출 결과

- 요구사항의 계산 결과 또는 제공 코드의 실행 결과
- Q1~Q4 답변

```python
# 문제 2-1. 두 방식은 각각 w=0에서 시작합니다.
X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])
initial_w = 0.0
learning_rate = 0.01

initial_grad_each = 2 * (initial_w * X - y) * X
batch_gradient = np.mean(initial_grad_each)
w_batch = initial_w - learning_rate * batch_gradient
batch_mse = np.mean((y - w_batch * X) ** 2)

# Mini-Batch 1: 입력 [1, 2], 정답 [2, 4]
first_grad_each = 2 * (initial_w * X[:2] - y[:2]) * X[:2]
first_gradient = np.mean(first_grad_each)
w_mini_first = initial_w - learning_rate * first_gradient
# Mini-Batch 2는 첫 갱신에서 얻은 w를 사용합니다.
second_grad_each = 2 * (w_mini_first * X[2:] - y[2:]) * X[2:]
second_gradient = np.mean(second_grad_each)
w_mini_second = w_mini_first - learning_rate * second_gradient

print("초기 데이터별 기울기:", initial_grad_each)
print("Batch 평균 기울기:", batch_gradient)
print(f"Batch 새 w={w_batch:.6f}, 전체 MSE={batch_mse:.6f}")
print("Mini-Batch 첫 묶음 기울기:", first_grad_each)
print(f"첫 평균 기울기={first_gradient:.6f}, 새 w={w_mini_first:.6f}")
print("Mini-Batch 두 번째 묶음 기울기:", np.round(second_grad_each, 6))
print(f"두 번째 평균 기울기={second_gradient:.6f}, 새 w={w_mini_second:.6f}")
print("한 Epoch의 업데이트 횟수: Batch=1, SGD=4, Mini-Batch(크기 2)=2")
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 초기 w=0의 데이터별 기울기는 `[−4, −16, −36, −64]`이고 평균은 $(-4-16-36-64)/4=-30$입니다.
- Batch 새 w는 $0-0.01(-30)=0.3$입니다. 제곱오차는 `[2.89, 11.56, 26.01, 46.24]`이므로 전체 MSE는 $86.7/4=21.675$입니다.
- Mini-Batch 첫 묶음은 기울기 `[−4, −16]`, 평균 −10이므로 새 w는 0.1입니다.
- 두 번째 묶음은 w=0.1에서 기울기를 다시 구합니다. $2(0.1\times3-6)\times3=-34.2$, $2(0.1\times4-8)\times4=-60.8$로 평균 −47.5입니다. 새 w는 $0.1-0.01(-47.5)=0.575$입니다.
- 네 데이터의 한 Epoch에서 Batch는 1회, SGD는 4회, 크기 2의 Mini-Batch는 2회 업데이트합니다.

**Q1.** 초기 w=0에서 네 데이터의 기울기와 전체 평균은 얼마인가요?

**답변:** 초기 개별 기울기는 `[-4, -16, -36, -64]`이고 전체 평균은 −30입니다.

**Q2.** Batch로 한 번 업데이트한 w와 전체 MSE는 얼마인가요?

**답변:** Batch로 한 번 갱신한 w는 0.3이고, 그 상태에서 네 데이터 전체의 MSE는 21.675입니다.

**Q3.** Mini-Batch의 첫 묶음과 두 번째 묶음에서 평균 기울기와 새 w는 얼마인가요?

**답변:** 첫 묶음의 평균 기울기는 −10, 새 w는 0.1입니다. 두 번째 묶음은 갱신된 w=0.1에서 평균 기울기 −47.5를 구하므로 새 w는 0.575입니다.

**Q4.** 한 Epoch 동안 Batch, SGD, 크기 2의 Mini-Batch는 각각 몇 번 업데이트하나요?

**답변:** Batch는 전체 데이터를 한 번에 사용하므로 1회, SGD는 데이터 하나마다 갱신하므로 4회, 크기 2의 Mini-Batch는 두 묶음이므로 2회입니다.

### 문제 3-1. 두 번의 업데이트 복습하기

### 문제 설명

교안 연습문제 1의 기본 계산을 복습합니다. 모델은 ŷ=wx+b, x=3, y=12입니다. 처음 w=2, b=1이고 학습률은 0.02입니다.

### 제공 공식

- 예측 ŷ=wx+b, 손실 L=(y−ŷ)²
- g_w=2(ŷ−y)x, g_b=2(ŷ−y)
- 새 w=w−0.02×g_w, 새 b=b−0.02×g_b

### 요구사항

- 초기 예측·손실·기울기를 계산합니다.
- 정확히 두 번 업데이트하고 매번 새 예측과 손실을 확인합니다.
- 문제 1-1의 순서와 코드를 참고해 숫자를 바꿔 계산하면 됩니다.
- 큰 학습률 비교나 추가 모델 구현은 요구하지 않습니다.

#### 해석 질문

**Q1.** 초기 예측, 손실, [w,b] 순서의 기울기는 얼마인가요?

**Q2.** 첫 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**Q3.** 두 번째 기울기와 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**Q4.** 왜 두 번째 업데이트에 첫 기울기를 그대로 사용하면 안 되나요?

#### 제출 결과

- 요구사항의 계산 결과 또는 제공 코드의 실행 결과
- Q1~Q4 답변

```python
# 문제 3-1. 새 데이터와 학습률로 두 번 갱신합니다.
x, y = 3.0, 12.0
w, b = 2.0, 1.0
learning_rate = 0.02
initial_prediction = w * x + b
initial_loss = (y - initial_prediction) ** 2
initial_grad_w = 2 * (initial_prediction - y) * x
initial_grad_b = 2 * (initial_prediction - y)

w1 = w - learning_rate * initial_grad_w
b1 = b - learning_rate * initial_grad_b
prediction1 = w1 * x + b1
loss1 = (y - prediction1) ** 2
grad_w1 = 2 * (prediction1 - y) * x
grad_b1 = 2 * (prediction1 - y)
w2 = w1 - learning_rate * grad_w1
b2 = b1 - learning_rate * grad_b1
prediction2 = w2 * x + b2
loss2 = (y - prediction2) ** 2

print("초기 예측, 손실:", initial_prediction, initial_loss)
print("초기 기울기 [w, b]:", np.round([initial_grad_w, initial_grad_b], 6))
print("1회 후 [w, b, 예측, 손실]:", np.round([w1, b1, prediction1, loss1], 6))
print("두 번째에 사용할 기울기:", np.round([grad_w1, grad_b1], 6))
print("2회 후 [w, b, 예측, 손실]:", np.round([w2, b2, prediction2, loss2], 6))
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 초기 예측은 $2\times3+1=7$, 손실은 $(12-7)^2=25$입니다. 기울기는 $g_w=2(7-12)\times3=-30$, $g_b=-10$입니다.
- 1회 후 $w=2-0.02(-30)=2.6$, $b=1-0.02(-10)=1.2$입니다. 예측은 9, 손실은 9입니다.
- 두 번째 기울기는 $g_w=2(9-12)\times3=-18$, $g_b=2(9-12)=-6$입니다.
- 2회 후 $w=2.6-0.02(-18)=2.96$, $b=1.2-0.02(-6)=1.32$입니다. 예측은 $2.96\times3+1.32=10.2$, 손실은 $(12-10.2)^2=3.24$입니다.

**Q1.** 초기 예측, 손실, [w,b] 순서의 기울기는 얼마인가요?

**답변:** 초기 예측은 7, 손실은 25, [w,b] 순서의 기울기는 `[-30, -10]`입니다.

**Q2.** 첫 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**답변:** w=2.6, b=1.2, 예측=9, 손실=9입니다.

**Q3.** 두 번째 기울기와 업데이트 후 w, b, 예측, 손실은 얼마인가요?

**답변:** 두 번째 기울기는 `[-18, -6]`입니다. 갱신 후 w=2.96, b=1.32, 예측=10.2, 손실=3.24입니다.

**Q4.** 왜 두 번째 업데이트에 첫 기울기를 그대로 사용하면 안 되나요?

**답변:** 첫 갱신으로 파라미터와 예측값이 달라져 현재 손실의 기울기도 달라졌기 때문입니다. 두 번째에는 새 상태의 기울기를 사용해야 합니다.

### 문제 3-2. 검증 손실 표로 학습 종료 시점 찾기

### 문제 설명

교안 연습문제 3의 손실 표를 읽습니다. 학습 손실은 학습 데이터의 손실이고, 검증 손실은 따로 둔 검증 데이터의 손실입니다. 이 문제에서는 검증 손실이 작을수록 좋은 상태로 판단합니다.

| Epoch | 학습 손실 | 검증 손실 |
|---|---:|---:|
| 1 | 0.80 | 0.60 |
| 2 | 0.65 | 0.45 |
| 3 | 0.50 | 0.40 |
| 4 | 0.40 | 0.42 |
| 5 | 0.32 | 0.43 |
| 6 | 0.25 | 0.44 |

### 사용할 규칙

- 지금까지 가장 작은 검증 손실보다 **더 작아졌을 때** 개선으로 봅니다.
- 개선하면 ‘연속 미개선 횟수’를 0으로 되돌립니다.
- 개선하지 않으면 그 횟수를 1 더합니다.
- patience=3: 연속으로 3개 Epoch에서 개선되지 않으면 그 Epoch가 끝난 뒤 학습을 멈춥니다.
- Epoch 1은 첫 기준값으로 기록하며, 그때의 연속 미개선 횟수는 0입니다.

### 요구사항

- 검증 손실이 가장 작은 Epoch과 그 값을 찾습니다.
- Epoch 4, 5, 6의 연속 미개선 횟수를 셉니다.
- 학습을 멈추는 Epoch과 검증 손실이 가장 좋았던 Epoch을 구별합니다.
- 표를 읽어 답하면 됩니다. 실제 모델 학습이나 Early Stopping 프로그램 작성은 필수가 아닙니다.

**Q1.** 검증 손실이 가장 작은 Epoch과 그 손실은 얼마인가요?

**Q2.** Epoch 4, 5, 6의 연속 미개선 횟수는 각각 얼마인가요?

**Q3.** patience=3이면 언제 학습을 멈추며, 가장 좋은 모델의 시점은 언제인가요?

**Q4.** 학습 손실이 계속 줄어도 이 규칙에서는 왜 학습을 멈추나요?

#### 제출 결과

- 요구사항의 계산 결과 또는 제공 코드의 실행 결과
- Q1~Q4 답변

```python
# 문제 3-2. 주어진 손실 표에서 개선 여부와 종료 시점을 확인합니다.
epochs = np.array([1, 2, 3, 4, 5, 6])
train_losses = np.array([0.80, 0.65, 0.50, 0.40, 0.32, 0.25])
validation_losses = np.array([0.60, 0.45, 0.40, 0.42, 0.43, 0.44])
patience = 3
best_loss = np.inf
best_epoch = None
wait = 0
stop_epoch = None
non_improvement_counts = []

print("Epoch / 학습 손실 / 검증 손실 / 연속 미개선 횟수")
for epoch, train_loss, val_loss in zip(epochs, train_losses, validation_losses):
    if val_loss < best_loss:
        best_loss = val_loss
        best_epoch = int(epoch)
        wait = 0
    else:
        wait += 1
    non_improvement_counts.append(wait)
    print(f"{epoch} / {train_loss:.2f} / {val_loss:.2f} / {wait}")
    if wait >= patience:
        stop_epoch = int(epoch)
        break

print(f"최적 검증 시점: Epoch {best_epoch}, 검증 손실 {best_loss:.2f}")
print(f"학습 종료 시점: Epoch {stop_epoch} 종료 후")
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 검증 손실은 Epoch 1~3에서 0.60 → 0.45 → 0.40으로 개선되므로 연속 미개선 횟수는 각각 0입니다.
- Epoch 4의 0.42, Epoch 5의 0.43, Epoch 6의 0.44는 지금까지의 최저값 0.40보다 작지 않으므로 횟수가 1 → 2 → 3이 됩니다.
- patience=3에 도달하는 Epoch 6이 끝난 뒤 멈춥니다. 가장 좋은 검증 상태는 Epoch 3이며, 종료 시점과 다릅니다.

**Q1.** 검증 손실이 가장 작은 Epoch과 그 손실은 얼마인가요?

**답변:** Epoch 3이며 검증 손실은 0.40입니다.

**Q2.** Epoch 4, 5, 6의 연속 미개선 횟수는 각각 얼마인가요?

**답변:** Epoch 4는 1회, Epoch 5는 2회, Epoch 6은 3회입니다.

**Q3.** patience=3이면 언제 학습을 멈추며, 가장 좋은 모델의 시점은 언제인가요?

**답변:** Epoch 6이 끝난 뒤 학습을 멈춥니다. 가장 좋은 모델의 시점은 검증 손실이 가장 작았던 Epoch 3입니다.

**Q4.** 학습 손실이 계속 줄어도 이 규칙에서는 왜 학습을 멈추나요?

**답변:** 이 규칙은 학습 손실이 아니라 검증 손실의 개선 여부를 기준으로 하기 때문입니다. 학습 데이터의 손실은 줄어도 검증 손실은 세 번 연속 개선되지 않아, 검증 성능이 더 좋아지지 않는 학습을 중단합니다.

### 실습 마무리 답변

**1.** **답변:** 각 파라미터에 대해 `새 값 = 현재 값 - 학습률 × 현재 손실 기울기`입니다. 벡터로는 $\theta_{\mathrm{new}}=\theta_{\mathrm{old}}-\alpha\nabla L$입니다.

**2.** **답변:** 양수 학습률에 음수 기울기를 곱한 값을 빼면 파라미터가 증가합니다. 예를 들어 $3-0.05(-12)=3.6$입니다.

**3.** **답변:** 같은 값이 아닙니다. 실제 변화량은 `-학습률 × 기울기`이므로 현재 기울기의 부호와 크기에도 영향을 받습니다.

**4.** **답변:** Epoch는 전체 학습 데이터를 한 번 사용한 단위이고, Iteration은 파라미터를 한 번 업데이트한 단위입니다. 여러 묶음을 사용하면 한 Epoch에 여러 Iteration이 있습니다.

**5.** **답변:** 정확히 0이라는 뜻은 아닙니다. 파라미터나 손실이 어떤 값에 가까워지는 현상이며, 이 예제도 30회 후 손실은 0에 가깝지만 양수입니다.

**6.** **답변:** 항상 같지 않습니다. 이번 표에서는 최적 검증 시점이 Epoch 3이고, patience=3을 기다린 종료 시점은 Epoch 6이 끝난 뒤입니다.
