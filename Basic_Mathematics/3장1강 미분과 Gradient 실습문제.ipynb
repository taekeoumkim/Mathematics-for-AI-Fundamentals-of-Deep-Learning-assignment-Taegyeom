## 먼저 읽는 용어와 기호

| 용어·기호 | 쉬운 뜻 |
|---|---|
| 함수 f(x) | x라는 입력을 넣으면 계산 결과를 내는 규칙. f(x)=x²라면 입력을 제곱함 |
| 변수 / 상수 | 값을 바꿀 수 있는 문자 / 지금 계산에서 고정된 값 |
| 대입 | 식의 문자 자리에 주어진 숫자를 넣는 일 |
| 제곱 x² / 코드 `x ** 2` | x를 두 번 곱함. 3²=3×3=9 |
| 2x / 코드 `2 * x` | 2와 x를 곱함. 코드에서는 곱셈 기호를 생략하지 않음 |
| 평균 변화율 | 출력이 변한 양을 입력이 변한 양으로 나눈 값 |

| ŷ / y | 예측값 / 실제 정답. ŷ는 ‘와이 햇’으로 읽음 |

| 손실 L | 예측과 정답의 차이를 평가한 값. 이 자료는 차이를 제곱한 손실을 사용 |
| 편미분 ∂L/∂w | 다른 변수는 고정하고 w만 아주 조금 바꿀 때의 순간 변화율 |
| Gradient ∇L | 여러 편미분값을 정해진 순서대로 묶은 벡터. ∇는 ‘나블라’로 읽음 |
| 벡터 [a,b] | 순서가 있는 숫자 묶음. 이 자료의 Gradient 순서는 항상 [w 방향, b 방향] |
| −∇L | Gradient의 각 성분에 −1을 곱해 방향을 뒤집은 벡터 |
| ≈ | 정확히 같은 값은 아니지만 대략 같다는 표시 |

**처음 보는 수식 읽기:** $\frac{f(x+h)-f(x)}{h}$에서 가로선 위가 분자, 아래가 분모입니다. 먼저 두 함수값을 빼고 그 결과를 h로 나눕니다. `∂L/∂w`는 이번 실습에서 L을 w로 단순히 나누라는 뜻이 아니라 **w에 대한 편미분을 나타내는 기호**입니다.

### 이번 실습에 사용하는 미분 규칙

| 함수의 항 | 미분한 결과 | 읽는 방법 |
|---|---|---|
| 상수 c | 0 | 입력이 달라도 상수는 변하지 않음 |
| ax | a | x의 계수가 기울기 |
| ax² | 2ax | 제곱항의 계수에 2를 곱하고 x를 남김 |
| 여러 항의 합 | 각 항을 미분한 결과의 합 | 항마다 따로 구한 뒤 더함 |

예를 들어 2x²+3x+5를 미분하면 4x+3입니다. 2x²는 4x, 3x는 3, 상수 5는 0이 됩니다. **함수값을 구하는 계산과 미분한 식을 만드는 계산은 다릅니다.** 먼저 어떤 것을 묻는지 확인하세요.

## 필수 실습 1. 함수의 변화와 미분

### 문제 1-1. 평균 변화율에서 미분값으로 연결하기

#### 문제 설명

함수는 $f(x)=x^2$입니다. 입력 x를 제곱한 값이 출력입니다. 먼저 구간 전체의 평균 변화율을 구하고, x=3에서 입력을 조금씩 바꿔 봅니다.

#### 요구사항

1. 구간 1→2, 2→3, 3→4, 4→5의 시작 함수값, 끝 함수값, 평균 변화율을 구하세요.
2. x=3으로 두고 h=1, 0.1, 0.01, 0.001일 때 아래 식을 계산하세요.
3. 결과가 어떤 수에 가까워지는지 확인하고 Q1~Q4에 답하세요.

$$\text{평균 변화율}=\frac{f(\text{끝 입력})-f(\text{시작 입력})}{\text{끝 입력}-\text{시작 입력}}$$

$$\frac{f(3+h)-f(3)}{h}$$

#### 작성 도움말

```python
def f(x):
    return x ** 2

starts = np.array([1, 2, 3, 4])
ends = np.array([2, 3, 4, 5])
rates = (f(ends) - f(starts)) / (ends - starts)
```

h는 `np.array([1, 0.1, 0.01, 0.001])`로 만들 수 있습니다. 교안의 미분 결과는 $f′(x)=2x$입니다. 이 식을 사용해 x=3에서의 미분값도 확인하세요. h 자리에 0을 직접 넣지 않습니다.

#### 해석 질문

**Q1.** 네 구간의 평균 변화율은 각각 얼마인가요?

**Q2.** h=1, 0.1, 0.01, 0.001일 때 변화율은 얼마이며 어떤 수에 가까워지나요?

**Q3.** f(3)과 f′(3)은 각각 얼마이며 뜻이 어떻게 다른가요?

**Q4.** 평균 변화율 식의 h에 0을 직접 넣어도 되나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 1-1. 구간의 평균 변화율과 한 점의 미분값을 비교합니다.
def f(x):
    return x ** 2

starts = np.array([1, 2, 3, 4])
ends = np.array([2, 3, 4, 5])
start_values = f(starts)
end_values = f(ends)
rates = (end_values - start_values) / (ends - starts)
print("구간 시작 입력:", starts)
print("구간 끝 입력:", ends)
print("시작 함수값:", start_values)
print("끝 함수값:", end_values)
print("구간별 평균 변화율:", rates)

x = 3.0
h_values = np.array([1, 0.1, 0.01, 0.001])
nearby_rates = (f(x + h_values) - f(x)) / h_values
derivative_at_x = 2 * x
print("h:", h_values)
print("x=3에서 h에 따른 변화율:", np.round(nearby_rates, 6))
print("함수값 f(3):", f(x))
print("미분값 f′(3):", derivative_at_x)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- 시작 함수값은 `[1, 4, 9, 16]`, 끝 함수값은 `[4, 9, 16, 25]`입니다. 각 구간의 입력 변화량은 1이므로 평균 변화율은 `[3, 5, 7, 9]`입니다.
- $((3+h)^2-3^2)/h=(6h+h^2)/h=6+h$이므로, h=1, 0.1, 0.01, 0.001에서 각각 7, 6.1, 6.01, 6.001입니다.
- h가 0에 가까워지면 변화율은 6에 가까워집니다. 주어진 도함수 $f′(x)=2x$에 3을 넣어도 6입니다.

**Q1.** 네 구간의 평균 변화율은 각각 얼마인가요?

**답변:** 1→2에서는 $(4-1)/(2-1)=3$, 2→3에서는 $(9-4)/(3-2)=5$, 3→4에서는 7, 4→5에서는 9입니다.

**Q2.** h=1, 0.1, 0.01, 0.001일 때 변화율은 얼마이며 어떤 수에 가까워지나요?

**답변:** 차례로 7, 6.1, 6.01, 6.001입니다. h가 작아질수록 x=3에서의 미분값인 6에 가까워집니다.

**Q3.** f(3)과 f′(3)은 각각 얼마이며 뜻이 어떻게 다른가요?

**답변:** $f(3)=9$는 입력 3에서의 함수 출력이고, $f′(3)=6$은 그 점에서 입력을 아주 조금 바꿀 때의 순간 변화율입니다. 함수의 높이와 기울기는 서로 다른 값입니다.

**Q4.** 평균 변화율 식의 h에 0을 직접 넣어도 되나요?

**답변:** 직접 넣을 수 없습니다. 분모가 0이 되어 원래 식이 0/0 꼴이 됩니다. 0이 아닌 h를 점점 줄였을 때 가까워지는 값으로 미분값을 이해합니다.

### 문제 1-2. 손실값과 미분값을 구분하기

#### 문제 설명

모델은 $\hat y=wx$이며 입력 x=2, 실제 정답 y=10입니다. w가 바뀌면 예측값과 손실이 달라집니다.

$$L(w)=(10-2w)^2=4w^2-40w+100$$
$$L′(w)=8w-40$$

#### 요구사항

1. w=2, 3, 4, 5, 6에 대해 예측값 2w, 손실값 L(w), 미분값 L′(w)를 계산하세요.
2. w=3과 w=6에서 손실을 줄이려면 w를 어느 쪽으로 조금 바꿔야 하는지 읽어 보세요.
3. Q1~Q4에 답하세요.

#### 작성 도움말

```python
w_values = np.array([2, 3, 4, 5, 6])
predictions = 2 * w_values
losses = (10 - predictions) ** 2
derivatives = 8 * w_values - 40
```

전개식과 미분식은 위 식을 사용하세요. `−40w`를 미분하면 `−40`입니다. 함수의 손실값과 미분값을 같은 숫자로 생각하지 않도록 따로 출력하세요.

#### 해석 질문

**Q1.** w=2,3,4,5,6의 예측값과 손실값은 각각 무엇인가요?

**Q2.** 같은 다섯 w에서 미분값은 각각 얼마인가요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 1-2. 함수의 손실값과 손실의 기울기를 따로 구합니다.
w_values = np.array([2, 3, 4, 5, 6])
predictions = 2 * w_values
losses = (10 - predictions) ** 2
derivatives = 8 * w_values - 40

print("w:", w_values)
print("예측값:", predictions)
print("손실값 L(w):", losses)
print("미분값 L′(w):", derivatives)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- w=2, 3, 4, 5, 6에서 예측값 2w는 `[4, 6, 8, 10, 12]`입니다.
- 정답 10과의 차이를 제곱하면 손실은 `[36, 16, 4, 0, 4]`입니다.
- $L′(w)=8w-40$에 대입하면 미분값은 `[-24, -16, -8, 0, 8]`입니다.
- w=3에서는 음수 기울기이므로 w를 조금 늘리고, w=6에서는 양수 기울기이므로 w를 조금 줄이면 손실이 감소합니다.

**Q1.** w=2,3,4,5,6의 예측값과 손실값은 각각 무엇인가요?

**답변:** w=2,3,4,5,6 순서로 예측값은 `[4, 6, 8, 10, 12]`, 손실값은 `[36, 16, 4, 0, 4]`입니다.

**Q2.** 같은 다섯 w에서 미분값은 각각 얼마인가요?

**답변:** 같은 순서로 `[-24, -16, -8, 0, 8]`입니다. 이는 손실값이 아니라 현재 w에서의 손실 변화율입니다.

**Q3.** w=3과 w=6에서 손실을 줄이려면 w를 어느 방향으로 조금 바꾸나요?

**답변:** w=3에서는 미분값이 −16이므로 w를 조금 늘립니다. w=6에서는 미분값이 +8이므로 w를 조금 줄입니다. 두 경우 모두 기울기와 반대쪽으로 조금 움직이는 것입니다.

**Q4.** w=5에서 손실이 최소인 이유를 이 문제의 손실식으로 설명하세요.

**답변:** $L(w)=(10-2w)^2$는 제곱이므로 항상 0 이상입니다. w=5에서 예측이 정답 10과 같아져 손실이 0이 되므로 이 손실식의 전역 최솟값입니다.

## 필수 실습 2. 편미분과 Gradient

### 문제 2-1. 한 변수만 움직이는 편미분 이해하기

#### 문제 설명

모델은 $\hat y=wx+b$, 입력 x=2, 실제 정답 y=10입니다. 현재 w=3, b=1입니다.

$$L(w,b)=(10-2w-b)^2$$
$$L(w,b)=4w^2+4wb+b^2-40w-20b+100$$
$$\frac{\partial L}{\partial w}=8w+4b-40,\qquad
\frac{\partial L}{\partial b}=4w+2b-20$$

#### 요구사항

1. 현재 예측값과 손실값을 계산하세요.
2. 제공된 식으로 w에 대한 편미분값과 b에 대한 편미분값을 구하세요.
3. **원래 점 (3,1)에서 각각 출발해** (3.1,1), (3,1.1)의 손실을 계산하세요. 앞의 변경을 다음 경우에 이어서 적용하지 않습니다.
4. 각 손실에서 원래 손실을 빼서 실제 변화량을 구하고 Q1~Q4에 답하세요.

#### 작성 도움말

```python
def loss(w, b):
    return (10 - 2 * w - b) ** 2
```

w만 바꾸는 경우는 `loss(3.1, 1)`, b만 바꾸는 경우는 `loss(3, 1.1)`입니다. 변화량은 **바꾼 뒤 손실 − 원래 손실**로 계산합니다. 편미분 식은 위 식을 그대로 사용해도 됩니다.

#### 해석 질문

**Q1.** 원래 점 (w,b)=(3,1)의 예측값과 손실값은 얼마인가요?

**Q4.** w에 대한 편미분값 −12를 “현재 손실이 −12”라고 읽어도 되나요?

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 2-1. 각 변경은 원래 점 (3, 1)에서 따로 시작합니다.
def loss(w, b):
    return (10 - 2 * w - b) ** 2

w, b = 3.0, 1.0
prediction = 2 * w + b
original_loss = loss(w, b)
partial_w = 8 * w + 4 * b - 40
partial_b = 4 * w + 2 * b - 20
loss_w_only = loss(w + 0.1, b)
loss_b_only = loss(w, b + 0.1)
change_w_only = loss_w_only - original_loss
change_b_only = loss_b_only - original_loss

print("원래 예측값:", prediction)
print("원래 손실:", original_loss)
print("w 편미분:", partial_w)
print("b 편미분:", partial_b)
print(f"w만 0.1 증가: 손실={loss_w_only:.4f}, 변화량={change_w_only:.4f}")
print(f"b만 0.1 증가: 손실={loss_b_only:.4f}, 변화량={change_b_only:.4f}")
````

**Q3.** w만 0.1 늘린 경우와 b만 0.1 늘린 경우의 손실 및 변화량은 각각 얼마인가요?

**Q4.** w에 대한 편미분값 −12를 “현재 손실이 −12”라고 읽어도 되나요?

### 문제 2-2. 편미분값을 묶어 Gradient 만들기

#### 문제 설명

앞 문제와 같은 손실입니다. $L(w,b)=(10-2w-b)^2$, 현재 점은 (3,1)입니다. 편미분값은 w 방향 −12, b 방향 −6입니다.

$$\nabla L=\left[\frac{\partial L}{\partial w},\frac{\partial L}{\partial b}\right]$$

#### 요구사항

3. **원래 점 (3,1)에서 각각 출발하여** 두 변수를 모두 0.1 늘린 (3.1,1.1), 모두 0.1 줄인 (2.9,0.9)의 손실과 실제 변화량을 구하세요.
4. Q1~Q4에 답하세요. Gradient의 크기, 단위벡터, 내적 근사는 계산하지 않습니다.

gradient = np.array([-12, -6])

손실은 문제 2-1의 함수를 사용하세요. 두 변수를 0.1씩 늘리는 [0.1,0.1]은 −∇L=[12,6]과 정확히 같은 방향은 아닙니다. 이번 비교는 교안에 제시된 **두 변수 동시 증가·감소 예제**입니다.

#### 해석 질문

**Q1.** 현재 Gradient는 무엇이며 각 성분은 어느 변수에 대응하나요?

**Q2.** −∇L은 무엇이며 Gradient와 어떤 관계인가요?

**Q3.** 두 변수를 모두 0.1 늘리거나 줄였을 때 손실과 변화량은 각각 얼마인가요?

**Q4.** 두 변수를 0.1씩 늘리는 이동을 정확한 −Gradient 방향이라고 부를 수 있나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 2-2. Gradient의 순서는 [w 방향, b 방향]입니다.
def loss(w, b):
    return (10 - 2 * w - b) ** 2

w, b = 3.0, 1.0
gradient = np.array([-12, -6])
negative_gradient = -gradient
original_loss = loss(w, b)
increased_loss = loss(w + 0.1, b + 0.1)
decreased_loss = loss(w - 0.1, b - 0.1)
increased_change = increased_loss - original_loss
decreased_change = decreased_loss - original_loss

print("Gradient [w, b]:", gradient)
print("-Gradient:", negative_gradient)
print("원래 손실:", original_loss)
print(f"두 변수 모두 0.1 증가: 손실={increased_loss:.4f}, 변화량={increased_change:.4f}")
print(f"두 변수 모두 0.1 감소: 손실={decreased_loss:.4f}, 변화량={decreased_change:.4f}")
````

#### 풀이 및 답변

**수식 및 계산 과정**

- [w 방향, b 방향] 순서의 Gradient는 `[-12, -6]`, −Gradient는 `[12, 6]`입니다.
- 원래 손실은 9입니다. 두 변수를 늘리면 $L(3.1,1.1)=(10-6.2-1.1)^2=7.29$이고 변화량은 −1.71입니다.
- 두 변수를 줄이면 $L(2.9,0.9)=(10-5.8-0.9)^2=10.89$이고 변화량은 +1.89입니다.
- 증가 이동 `[0.1, 0.1]`의 성분 비율은 1:1이고 −Gradient `[12, 6]`의 비율은 2:1이므로 정확히 같은 방향은 아닙니다.

**Q1.** 현재 Gradient는 무엇이며 각 성분은 어느 변수에 대응하나요?

**답변:** 현재 Gradient는 `[-12, -6]`입니다. 첫 성분 −12는 w 편미분, 둘째 성분 −6은 b 편미분에 대응합니다.

**Q2.** −∇L은 무엇이며 Gradient와 어떤 관계인가요?

**답변:** −Gradient는 `[12, 6]`입니다. Gradient의 모든 성분에 −1을 곱해 방향을 반대로 바꾼 벡터입니다.

**Q3.** 두 변수를 모두 0.1 늘리거나 줄였을 때 손실과 변화량은 각각 얼마인가요?

**답변:** 모두 0.1 늘리면 손실은 7.29, 실제 변화량은 −1.71입니다. 모두 0.1 줄이면 손실은 10.89, 실제 변화량은 +1.89입니다.

**Q4.** 두 변수를 0.1씩 늘리는 이동을 정확한 −Gradient 방향이라고 부를 수 있나요?

**답변:** 정확한 −Gradient 방향이라고 부를 수 없습니다. `[0.1, 0.1]`은 `[12, 6]`의 같은 양수 배수가 아니기 때문입니다. 이번 예제에서는 두 변수를 함께 늘리는 이동도 손실을 줄이지만 방향 자체는 다릅니다.

### 문제 3-1. 다항식의 미분과 값 대입 복습하기

3. Q1~Q4에 답하세요. 새로운 공식을 유도할 필요는 없습니다.

#### 작성 도움말

2x²를 미분하면 4x, −4x를 미분하면 −4, 상수 3을 미분하면 0입니다. 이 세 결과를 더하세요. `x_values = np.array([1, 2, 3])`으로 세 입력을 한 번에 계산할 수 있습니다. 함수값은 원래 식, 미분값은 미분한 식에 대입합니다.

#### 해석 질문

**Q1.** C′(x)는 무엇이며 각 항은 어떻게 미분되나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 3-1. 각 항을 미분하면 C′(x) = 4x - 4입니다.
def C(x):
    return 2 * x ** 2 - 4 * x + 3

def C_prime(x):
    return 4 * x - 4

x_values = np.array([1, 2, 3])
function_values = C(x_values)
derivative_values = C_prime(x_values)
print("C′(x) = 4x - 4")
print("x:", x_values)
print("함수값 C(x):", function_values)
print("미분값 C′(x):", derivative_values)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- $2x^2$의 미분은 4x, −4x의 미분은 −4, 상수 3의 미분은 0이므로 $C′(x)=4x-4$입니다.
- $C(1)=2-4+3=1$, $C(2)=8-8+3=3$, $C(3)=18-12+3=9$입니다.
- $C′(1)=0$, $C′(2)=4$, $C′(3)=8$입니다.

**Q1.** C′(x)는 무엇이며 각 항은 어떻게 미분되나요?

**답변:** $C′(x)=4x-4$입니다. 제곱항 2x²는 4x, 일차항 −4x는 −4, 상수 3은 0으로 미분합니다.

**Q2.** x=1,2,3의 함수값 C(x)는 각각 얼마인가요?

**답변:** x=1,2,3에서 함수값은 차례로 1, 3, 9입니다.

**Q3.** x=1,2,3의 미분값 C′(x)는 각각 얼마인가요?

**답변:** x=1,2,3에서 미분값은 차례로 0, 4, 8입니다.

**Q4.** C′(1)=0이면 C(1)도 0인가요?

**답변:** 그렇지 않습니다. $C′(1)=0$은 그 점의 순간 기울기가 0이라는 뜻이고, 실제 함수값은 $C(1)=1$입니다.

### 문제 3-2. 주어진 편미분식으로 Gradient 복습하기

#### 문제 설명

교안 연습문제 3의 모델을 사용합니다.

$$\hat y=wx+2bx,\qquad x=1,\ y=8,\ w=2,\ b=1$$

**이 문제의 b는 일반적인 ŷ=wx+b의 편향과 다릅니다.** 교안에 주어진 식 그대로 2bx의 계수입니다. x=1이므로 예측식은 w+2b가 됩니다.

$$L(w,b)=(8-w-2b)^2$$
$$L(w,b)=w^2+4wb+4b^2-16w-32b+64$$
$$\frac{\partial L}{\partial w}=2w+4b-16,\qquad
\frac{\partial L}{\partial b}=4w+8b-32$$

#### 요구사항

1. 현재 예측값과 손실값을 구하세요.
2. 제공된 두 편미분식에 w=2, b=1을 대입하세요.
3. [w 방향, b 방향] 순서의 Gradient와 −Gradient를 만드세요.
4. Q1~Q4에 답하세요. 벡터의 크기, 단위 방향, 내적 근사, 안장점 계산은 요구하지 않습니다.

#### 작성 도움말

예측은 `w * x + 2 * b * x`, 손실은 `(y - prediction) ** 2`입니다. 두 편미분값을 `np.array([partial_w, partial_b])`로 묶고 앞에 `-`를 붙이면 반대 방향입니다. 전개식과 미분식은 위 식을 참고하세요.

#### 해석 질문

**Q1.** 현재 예측값과 손실값은 얼마인가요?

**Q2.** w와 b에 대한 편미분값은 각각 얼마인가요?

**Q3.** Gradient와 −Gradient는 각각 무엇인가요?

**Q4.** 이번 모델을 앞 문제처럼 ŷ=2w+b로 계산해도 되나요?

#### 제출 결과

- 요구사항의 계산 결과
- Q1~Q4 답변

````python
# 문제 3-2. 이번 예측식은 wx + 2bx이며 일반적인 wx+b와 다릅니다.
x, y = 1.0, 8.0
w, b = 2.0, 1.0
prediction = w * x + 2 * b * x
loss_value = (y - prediction) ** 2
partial_w = 2 * w + 4 * b - 16
partial_b = 4 * w + 8 * b - 32
gradient = np.array([partial_w, partial_b])
negative_gradient = -gradient

print("예측값:", prediction)
print("손실값:", loss_value)
print("w 편미분:", partial_w)
print("b 편미분:", partial_b)
print("Gradient [w, b]:", gradient)
print("-Gradient:", negative_gradient)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- 예측은 $wx+2bx=2\times1+2\times1\times1=4$, 손실은 $(8-4)^2=16$입니다.
- w 편미분은 $2\times2+4\times1-16=-8$, b 편미분은 $4\times2+8\times1-32=-16$입니다.
- Gradient는 `[-8, -16]`, −Gradient는 `[8, 16]`입니다.

**Q1.** 현재 예측값과 손실값은 얼마인가요?

**답변:** 현재 예측값은 4이고 손실값은 16입니다.

**Q2.** w와 b에 대한 편미분값은 각각 얼마인가요?

**답변:** w에 대한 편미분은 −8, b에 대한 편미분은 −16입니다.

**Q3.** Gradient와 −Gradient는 각각 무엇인가요?

**답변:** [w,b] 순서의 Gradient는 `[-8, -16]`이고, −Gradient는 `[8, 16]`입니다.

**Q4.** 이번 모델을 앞 문제처럼 ŷ=2w+b로 계산해도 되나요?

**답변:** 그렇게 계산하면 다른 모델이 됩니다. 이번 식은 x=1에서 $\hat y=w+2b$이며, b는 일반적인 편향 항이 아니라 2bx에 들어가는 파라미터입니다. 주어진 식과 그 편미분식을 사용해야 합니다.
