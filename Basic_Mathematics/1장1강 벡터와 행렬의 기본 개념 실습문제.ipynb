### 문제 1-1. 스칼라·벡터·행렬과 Shape 확인

#### 문제 설명

교안에 나온 고객 3명의 활동 데이터입니다. 숫자 하나, 고객 한 명의 정보, 고객 전체의 정보를 각각 표현해 보세요.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 5 | 2 |
| B | 3 | 1 |
| C | 8 | 4 |

원소의 순서는 항상 **방문 횟수 → 장바구니 추가 횟수**, 고객의 순서는 **A → B → C**입니다.

#### 먼저 이해하기

- **행**은 가로줄이고 **열**은 세로줄입니다. 이 표에서는 고객 한 명을 한 행으로 옮깁니다.
- ‘원소’는 벡터나 행렬 안의 숫자 하나를 말합니다.
- `X₂₁`은 ‘X의 2행 1열’로 읽습니다. 아래의 작은 숫자는 곱할 숫자가 아니라 **위치**를 표시합니다. 앞은 행, 뒤는 열이고 이 수학 표기에서는 1부터 셉니다.
- 벡터의 차원은 ‘숫자가 몇 개인가’, `ndim`은 ‘축이 몇 개인가’, `shape`는 ‘각 축의 크기가 얼마인가’를 묻습니다.

#### 요구사항

1. 고객 A의 방문 횟수를 스칼라 `visit_a`에 저장하고 출력하세요.
2. 고객 A의 두 특성을 NumPy 배열 `x_a`로 만들고 출력하세요.
3. 전체 고객의 두 특성을 NumPy 배열 `X`로 만들고 출력하세요.
4. `x_a`와 `X`의 `ndim`, `shape`를 각각 출력하세요.
5. `x_a.reshape(1, 2)`로 행 벡터 형태 `x_row`, `x_a.reshape(2, 1)`로 열 벡터 형태 `x_col`을 만들고, 각각의 배열과 `shape`를 출력하세요.
6. 행렬의 원소 X₂₁, X₃₂의 값과 의미를 답변 작성란에 적으세요. 이 문항은 표를 보고 답해도 됩니다.

#### 작성 도움말

먼저 `visit_a = 표에서 찾은 숫자`처럼 값 하나를 저장하세요. 벡터는 `np.array([첫 특성값, 둘째 특성값])`, 행렬은 고객별 작은 대괄호를 다시 바깥 대괄호로 묶습니다.

`print(x_a.shape)`와 `print(x_a.ndim)`으로 결과를 확인하세요. `reshape(행 수, 열 수)`는 숫자의 값과 개수를 유지하면서 배치만 바꿉니다. 예를 들어 `reshape(1, 2)`는 1행 2열을 뜻합니다.

#### 해석 질문

**Q1.** `visit_a`, `x_a`, `X`는 각각 왜 스칼라, 벡터, 행렬인가요?

**Q2.** `x_a`는 수학적으로 몇 차원 벡터인가요? 그런데 `x_a.ndim`이 1인 이유는 무엇인가요?

**Q3.** `X.shape`의 두 숫자는 각각 무엇을 뜻하나요? X₂₁과 X₃₂는 어떤 값인가요?

**Q4.** `(2,)`, `(1, 2)`, `(2, 1)`의 차이는 무엇인가요? `x_a`를 `(2, 3)`으로 바꿀 수 있나요?

#### 제출 결과

- 스칼라·벡터·행렬 출력
- 각 배열의 ndim과 Shape, 행·열 벡터 출력
- X₂₁과 X₃₂의 값·의미 및 Q1~Q4 답변

```python
# 문제 1-1. 숫자 하나, 한 고객, 전체 고객을 각각 표현합니다.
visit_a = 5
x_a = np.array([5, 2])
X = np.array([[5, 2], [3, 1], [8, 4]])

print("visit_a (스칼라):", visit_a)
print("x_a (벡터):", x_a)
print("X (행렬):\n", X)
print("x_a.ndim:", x_a.ndim, "/ x_a.shape:", x_a.shape)
print("X.ndim:", X.ndim, "/ X.shape:", X.shape)

# 원소의 값과 개수를 유지하면서 행·열 배치만 바꿉니다.
x_row = x_a.reshape(1, 2)
x_col = x_a.reshape(2, 1)
print("x_row:\n", x_row, "\nshape:", x_row.shape)
print("x_col:\n", x_col, "\nshape:", x_col.shape)

# 수학의 행·열 번호는 1부터, NumPy 인덱스는 0부터 셉니다.
print("X₂₁ = X[1, 0]:", X[1, 0], "(고객 B의 방문 횟수)")
print("X₃₂ = X[2, 1]:", X[2, 1], "(고객 C의 장바구니 추가 횟수)")
```

#### 풀이 및 답변

**수식 및 계산 과정**

- `visit_a = 5`, `x_a = [5, 2]`, `X = [[5, 2], [3, 1], [8, 4]]`입니다.
- `x_a`: 원소 2개, `ndim = 1`, `shape = (2,)`입니다. `X`: 3행 2열, `ndim = 2`, `shape = (3, 2)`입니다.
- 행 벡터 형태는 `[[5, 2]]`, Shape는 `(1, 2)`이고, 열 벡터 형태는 `[[5], [2]]`, Shape는 `(2, 1)`입니다.
- 수학 표기의 $X_{21}=3$, $X_{32}=4$이며, NumPy에서는 각각 `X[1, 0]`, `X[2, 1]`로 읽습니다.

**Q1. `visit_a`, `x_a`, `X`는 각각 왜 스칼라, 벡터, 행렬인가요?**

- **답변:** `visit_a`는 숫자 하나이므로 스칼라입니다. `x_a`는 한 고객의 두 특성을 순서대로 모았으므로 벡터입니다. `X`는 여러 고객의 특성을 행과 열로 배치했으므로 행렬입니다.

**Q2. `x_a`는 수학적으로 몇 차원 벡터인가요? 그런데 `x_a.ndim`이 1인 이유는 무엇인가요?**

- **답변:** 원소가 2개이므로 수학적으로 2차원 벡터입니다. `ndim`은 원소 수가 아니라 배열의 축 수를 뜻합니다. `x_a`는 하나의 축을 가진 1차원 배열이므로 `x_a.ndim`은 1입니다.

**Q3. `X.shape`의 두 숫자는 각각 무엇을 뜻하나요? X₂₁과 X₃₂는 어떤 값인가요?**

- **답변:** `(3, 2)`의 3은 고객 3명에 해당하는 행 수이고, 2는 입력 특성 2개에 해당하는 열 수입니다. $X_{21}=3$은 고객 B의 방문 횟수, $X_{32}=4$는 고객 C의 장바구니 추가 횟수입니다.

**Q4. `(2,)`, `(1, 2)`, `(2, 1)`의 차이는 무엇인가요? `x_a`를 `(2, 3)`으로 바꿀 수 있나요?**

- **답변:** `(2,)`는 원소 2개의 1차원 배열로, 행·열 방향을 따로 갖지 않습니다. `(1, 2)`는 1행 2열, `(2, 1)`은 2행 1열인 2차원 배열입니다. `reshape`는 원소 수를 유지하므로, 원소 2개인 `x_a`를 원소 6개가 필요한 `(2, 3)`으로 바꿀 수 없습니다.

### 문제 1-2. 입력·정답·가중치·편향 구분과 점수 계산

#### 문제 설명

문제 1-1과 같은 고객 데이터에 실제 구매 여부를 추가했습니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 | 실제 구매 여부 |
|---|---:|---:|---:|
| A | 5 | 2 | 1 |
| B | 3 | 1 | 0 |
| C | 8 | 4 | 1 |

구매한 경우는 1, 구매하지 않은 경우는 0입니다. 방문 횟수의 가중치는 2, 장바구니 추가 횟수의 가중치는 3, 편향은 1로 정합니다.

교안의 계산식을 그대로 사용합니다.

$$\mathrm{score}=w_1x_1+w_2x_2+b$$

| 기호 | 의미 |
|---|---|
| $x_1$, $x_2$ | 방문 횟수, 장바구니 추가 횟수 |
| $w_1$, $w_2$ | 각 특성에 곱하는 가중치 |
| $b$ | 마지막에 더하는 편향 |
| score | 가중합으로 계산한 출력 점수 |

또한 교안의 두 출력용 가중치 표를 행렬로 표현해 보세요.

| 입력 특성 | 구매 관심도 가중치 | 고객 활동 가중치 |
|---|---:|---:|
| 방문 횟수 | 2 | 1 |
| 장바구니 추가 횟수 | 3 | 4 |

#### 수식을 말로 읽기

`w₁x₁`은 **첫 가중치 × 첫 입력값**입니다. 두 글자가 붙어 있어도 곱셈을 뜻합니다. 아래첨자 1과 2는 첫 번째·두 번째를 구분하는 번호입니다.

전체 식은 ‘방문 횟수에 해당 가중치를 곱한다 → 장바구니 추가 횟수에도 해당 가중치를 곱한다 → 두 결과와 편향을 더한다’는 뜻입니다. **가중합**은 이렇게 ‘각각 곱한 뒤 합한 값’입니다. **편향**은 그 합에 마지막으로 더하는 값입니다.

#### 요구사항

1. 방문 횟수와 장바구니 추가 횟수만 포함하는 입력 행렬 `X`, 실제 구매 여부를 담은 타깃 벡터 `y`를 만들고 출력하세요.

4. 두 출력용 표를 가중치 행렬 `W`로 만들고 배열과 Shape를 출력하세요. 행은 입력 특성, 열은 출력 점수에 대응하도록 배치하세요. 이 문항에서는 W를 이용한 출력 계산은 하지 않습니다.

#### 작성 도움말

교안의 수식에 표의 숫자를 직접 넣어 `가중치 * 입력값 + 가중치 * 입력값 + 편향` 순서로 적으세요. 코드에서 곱셈은 `*`입니다. 입력과 가중치의 짝을 표에서 먼저 확인하세요.

```python
# 문제 1-2. 입력과 실제 정답을 분리하고 모델의 값을 지정합니다.
X = np.array([[5, 2], [3, 1], [8, 4]])
y = np.array([1, 0, 1])
w = np.array([2, 3])
b = 1

print("X:\n", X, "\nshape:", X.shape)
print("y:", y, "/ shape:", y.shape)
print("w:", w, "/ shape:", w.shape)
print("b:", b)

x_a = np.array([5, 2])
# 방문 횟수에는 2, 장바구니 추가 횟수에는 3을 곱합니다.
score_a = 2 * 5 + 3 * 2 + 1
print("x_a:", x_a)
print("score_a = 2 * 5 + 3 * 2 + 1 =", score_a)

# 행: 입력 특성, 열: 구매 관심도·고객 활동 점수
W = np.array([[2, 1], [3, 4]])
print("W:\n", W, "\nshape:", W.shape)
```

### 문제 2-1. 음수가 포함된 벡터의 L1 Norm과 L2 Norm

#### 문제 설명

교안의 벡터 `v = [3, -4]`를 사용합니다. 이 벡터는 노름 계산을 연습하기 위한 숫자 예시입니다.

노름은 벡터의 크기를 하나의 숫자로 표현합니다. 두 원소에 대한 공식은 다음과 같습니다.

$$\|v\|_1=|v_1|+|v_2|$$
$$\|v\|_2=\sqrt{v_1^2+v_2^2}$$

- $v_1$, $v_2$: 벡터의 첫 번째와 두 번째 원소
- $|a|$: a의 절댓값. 예를 들어 −4의 절댓값은 4입니다.
- $a^2$: a를 자기 자신과 곱한 값
- $\sqrt{a}$: 제곱하여 a가 되는 수 중 음수가 아닌 값. 예를 들어 $\sqrt{25}=5$입니다.
- $\|v\|_1$, $\|v\|_2$: 각각 벡터 v의 L1 Norm, L2 Norm

#### 수식을 말로 읽기

- `||v||₁`은 ‘벡터 v의 엘원 노름’, `||v||₂`는 ‘벡터 v의 엘투 노름’으로 읽습니다. 아래의 1과 2는 **계산 방법의 종류**입니다.
- 절댓값은 숫자가 0에서 얼마나 떨어져 있는지 나타냅니다. 양수는 그대로이고 음수는 크기만 남깁니다.
- 제곱은 같은 수를 두 번 곱하는 것입니다. 음수 전체를 제곱할 때는 `(-4)²`처럼 괄호로 묶어 읽으세요.
- 제곱근 기호 √는 ‘어떤 음수가 아닌 수를 두 번 곱해야 이 값이 되는가’를 묻습니다.

#### 요구사항

1. `v = [3, -4]`를 NumPy 배열로 만들고 출력하세요.
2. `np.linalg.norm(v, ord=1)`로 L1 Norm을 계산하고 출력하세요.
3. `np.linalg.norm(v, ord=2)`로 L2 Norm을 계산하고 출력하세요.
4. 답변 작성란에 두 공식의 대입·계산 과정을 적고 코드 결과와 비교하세요.

#### 작성 도움말

`ord=1`은 절댓값의 합, `ord=2`는 제곱합의 제곱근을 계산하도록 선택합니다.

#### 해석 질문

**Q1.** L1 Norm은 얼마인가요? 각 원소의 절댓값부터 계산 과정을 적으세요.

**Q2.** L2 Norm은 얼마인가요? 각 원소의 제곱, 합, 제곱근 순서로 계산 과정을 적으세요.

**Q3.** 원소를 그냥 더한 `3 + (-4)`와 L1 Norm이 다른 이유는 무엇인가요?

**Q4.** 원소가 2개라는 사실과 L2 Norm의 계산 결과는 같은 뜻인가요?

#### 제출 결과

- v와 두 노름 출력
- 공식에 값을 대입한 계산 과정 및 Q1~Q4 답변

```python
# 문제 2-1. 음수도 절댓값 또는 제곱을 거쳐 크기에 반영합니다.
v = np.array([3, -4])
v_l1 = np.linalg.norm(v, ord=1)
v_l2 = np.linalg.norm(v, ord=2)

print("v:", v)
print("L1 Norm = |3| + |-4| =", v_l1)
print("L2 Norm = sqrt(3² + (-4)²) =", v_l2)
```

#### 풀이 및 답변

**수식 및 계산 과정**

- L1 Norm: $\|v\|_1=|3|+|-4|=3+4=7$.
- L2 Norm: $\|v\|_2=\sqrt{3^2+(-4)^2}=\sqrt{9+16}=\sqrt{25}=5$.
- NumPy 결과인 `7.0`, `5.0`과 손계산이 일치합니다.

**Q1. L1 Norm은 얼마인가요? 각 원소의 절댓값부터 계산 과정을 적으세요.**

- **답변:** L1 Norm은 7입니다. 첫 원소의 절댓값은 $|3|=3$, 둘째는 $|-4|=4$이므로, 두 값을 더하면 $3+4=7$입니다.

**Q2. L2 Norm은 얼마인가요? 각 원소의 제곱, 합, 제곱근 순서로 계산 과정을 적으세요.**

- **답변:** L2 Norm은 5입니다. $3^2=9$, $(-4)^2=16$이고, 합은 25이므로 제곱근은 $\sqrt{25}=5$입니다.

**Q3. 원소를 그냥 더한 `3 + (-4)`와 L1 Norm이 다른 이유는 무엇인가요?**

- **답변:** 그냥 더하면 양수와 음수가 상쇄되어 $3+(-4)=-1$이 됩니다. L1 Norm은 각 원소의 절댓값을 더하므로 부호가 상쇄되지 않아 7이 됩니다.

**Q4. 원소가 2개라는 사실과 L2 Norm의 계산 결과는 같은 뜻인가요?**

- **답변:** 같은 뜻이 아닙니다. 원소 수 2는 벡터의 수학적 차원이고, L2 Norm 5는 원소의 값으로 계산한 벡터의 길이입니다.

### 문제 2-2. 두 고객 사이의 거리와 벡터 정규화

#### 문제 설명

교안의 연습문제와 같은 두 고객의 활동 데이터입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 4 |
| B | 6 | 8 |

원소 순서는 방문 횟수 → 장바구니 추가 횟수입니다.

두 고객 사이의 거리는 **B 벡터 − A 벡터**에 노름을 적용합니다.

$$d_1(A,B)=\|x_B-x_A\|_1,\qquad d_2(A,B)=\|x_B-x_A\|_2$$

길이가 1인 벡터를 만들 때는 각 벡터를 **자신의 L2 Norm**으로 나눕니다.

$$x_{\mathrm{normalized}}=\frac{x}{\|x\|_2}$$

이 방법은 L2 Norm이 0이 아닌 벡터에 적용합니다.

#### 수식을 말로 읽기

`x_B−x_A`는 **같은 위치의 특성끼리 뺀다**는 뜻입니다. 방문 횟수끼리, 장바구니 추가 횟수끼리 뺍니다. `d₁(A,B)`와 `d₂(A,B)`는 A와 B 사이의 거리를 두 방법으로 계산한 값입니다.

정규화 식의 분수는 ‘위에 있는 벡터의 **모든 원소**를 아래의 L2 Norm 값으로 나눈다’고 읽습니다. 길이가 1이라는 뜻은 **L2 Norm이 1**이라는 뜻입니다.

**1~3번은 거리**, **4~5번은 원래 길이를 구한 뒤 정규화**, **6~7번은 결과 확인**입니다. 한 묶음씩 진행하세요.

#### 요구사항

1. 두 고객의 특성을 각각 `x_a`, `x_b`에 NumPy 배열로 저장하세요.
2. 차이 벡터 `diff = x_b - x_a`를 계산하고 출력하세요.
3. 차이 벡터의 L1 Norm과 L2 Norm으로 맨해튼 거리와 유클리드 거리를 계산하고 출력하세요.
4. A, B 각 벡터의 L2 Norm을 `a_l2`, `b_l2`에 저장하고 출력하세요.
5. 각 벡터를 자신의 L2 Norm으로 나누어 `a_normalized`, `b_normalized`를 만들고 출력하세요.
6. 정규화된 두 벡터의 L2 Norm을 각각 출력하세요.
7. 교안의 `np.allclose(a_normalized, b_normalized)`를 그대로 사용해 결과를 비교하세요. 이 함수는 두 배열의 같은 위치 값들이 작은 계산 오차 범위 안에서 같은지 확인하며, `True`이면 비교 조건을 만족한다는 뜻입니다.

#### 작성 도움말

`np.linalg.norm(diff, ord=1)`처럼 차이 벡터를 넣으면 거리이고, `np.linalg.norm(x_a, ord=2)`처럼 원래 벡터를 넣으면 해당 벡터의 크기입니다.

#### 해석 질문

**Q1.** 차이 벡터와 두 거리의 값은 무엇인가요? 거리 계산 전에 차이 벡터를 만드는 이유는 무엇인가요?

**Q2.** A 벡터의 L2 Norm과 A·B 사이의 유클리드 거리는 각각 무엇을 측정하나요?

**Q3.** A, B의 정규화 계산 과정을 적으세요. 정규화 후 L2 Norm은 각각 얼마인가요?

**Q4.** 정규화 결과가 같다는 것은 두 고객의 원래 활동 횟수도 같다는 뜻인가요?

**Q5.** `[0, 0]`을 같은 방법으로 정규화하면 안 되는 이유는 무엇인가요?

#### 제출 결과

- 차이 벡터, 맨해튼·유클리드 거리
- 정규화 전후 L2 Norm, 정규화 벡터와 비교 결과
- Q1~Q5 답변

```python
# 문제 2-2. 두 고객 사이의 거리부터 계산합니다.
x_a = np.array([3, 4])
x_b = np.array([6, 8])
diff = x_b - x_a
manhattan_distance = np.linalg.norm(diff, ord=1)
euclidean_distance = np.linalg.norm(diff, ord=2)

print("x_a:", x_a, "/ x_b:", x_b)
print("diff = x_b - x_a:", diff)
print("맨해튼 거리:", manhattan_distance)
print("유클리드 거리:", euclidean_distance)

# 각 벡터를 자신의 L2 Norm으로 나눕니다. 두 노름은 모두 0이 아닙니다.
a_l2 = np.linalg.norm(x_a, ord=2)
b_l2 = np.linalg.norm(x_b, ord=2)
a_normalized = x_a / a_l2
b_normalized = x_b / b_l2

print("정규화 전 A의 L2 Norm:", a_l2)
print("정규화 전 B의 L2 Norm:", b_l2)
print("a_normalized:", a_normalized)
print("b_normalized:", b_normalized)
print("정규화 후 A의 L2 Norm:", np.linalg.norm(a_normalized, ord=2))
print("정규화 후 B의 L2 Norm:", np.linalg.norm(b_normalized, ord=2))
print("정규화 결과 비교:", np.allclose(a_normalized, b_normalized))
```

#### 풀이 및 답변

**수식 및 계산 과정**

- 차이 벡터는 `diff = [6 - 3, 8 - 4] = [3, 4]`입니다.
- 맨해튼 거리: $|3|+|4|=7$. 유클리드 거리: $\sqrt{3^2+4^2}=5$.
- 원래 길이는 $\|x_A\|_2=5$, $\|x_B\|_2=\sqrt{6^2+8^2}=10$입니다.
- A 정규화: `[3 / 5, 4 / 5] = [0.6, 0.8]`. B 정규화: `[6 / 10, 8 / 10] = [0.6, 0.8]`.
- 두 벡터 모두 정규화 후 L2 Norm은 $\sqrt{0.6^2+0.8^2}=1$이며, `np.allclose(...)`의 결과는 `True`입니다.

**Q1. 차이 벡터와 두 거리의 값은 무엇인가요? 거리 계산 전에 차이 벡터를 만드는 이유는 무엇인가요?**

- **답변:** 차이 벡터는 `[3, 4]`, 맨해튼 거리는 7, 유클리드 거리는 5입니다. 거리에서는 각 특성이 서로 얼마나 다른지를 측정해야 하므로, 같은 위치의 값끼리 뺀 차이 벡터를 먼저 만듭니다.

**Q2. A 벡터의 L2 Norm과 A·B 사이의 유클리드 거리는 각각 무엇을 측정하나요?**

- **답변:** A의 L2 Norm은 원점에서 A까지의 길이이고, A·B 사이의 유클리드 거리는 두 고객의 위치 사이의 길이입니다. 이 예제에서는 둘 다 5이지만 측정하는 대상은 다릅니다.

**Q3. A, B의 정규화 계산 과정을 적으세요. 정규화 후 L2 Norm은 각각 얼마인가요?**

- **답변:** A는 자신의 길이 5로 나누어 `[0.6, 0.8]`, B는 자신의 길이 10으로 나누어 `[0.6, 0.8]`이 됩니다. 정규화 후 두 벡터의 L2 Norm은 각각 1입니다.

**Q4. 정규화 결과가 같다는 것은 두 고객의 원래 활동 횟수도 같다는 뜻인가요?**

- **답변:** 원래 활동 횟수가 같다는 뜻은 아닙니다. B의 방문·장바구니 횟수는 A의 2배입니다. 두 벡터의 방향과 특성 간 비율이 같아 길이를 1로 맞춘 결과가 같습니다.

**Q5. `[0, 0]`을 같은 방법으로 정규화하면 안 되는 이유는 무엇인가요?**

- **답변:** `[0, 0]`의 L2 Norm은 0이므로 각 원소를 0으로 나누게 됩니다. 이 나눗셈은 정의되지 않으므로 같은 공식으로 정규화할 수 없습니다.

## 과제. 같은 방법을 다른 숫자에 적용하기

### 문제 3-1. 입력·정답 표현과 고객 C의 점수 계산

#### 문제 설명

교안 **연습문제 2**의 데이터입니다. 필수 1-2와 같은 방법으로 표를 옮기고 고객 C의 점수를 계산하세요.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 | 실제 구매 여부 |
|---|---:|---:|---:|
| A | 4 | 2 | 1 |
| B | 6 | 1 | 0 |
| C | 3 | 3 | 1 |

특성 순서는 방문 횟수 → 장바구니 추가 횟수입니다. 가중치는 각각 2와 3, 편향은 1입니다. 실제 구매 여부의 1은 구매함, 0은 구매하지 않음입니다.

`score = w₁×x₁ + w₂×x₂ + b`

식은 ‘각 특성에 가중치를 곱하고, 두 곱셈 결과와 편향을 더한다’고 읽습니다.

#### 요구사항

3. X, y, w의 Shape를 각각 출력하세요.
4. 고객 C의 벡터 `x_c`를 만들고, 식에 숫자를 직접 넣은 곱셈·덧셈으로 `score_c`를 계산하여 출력하세요.

#### 작성 도움말

필수 1-2를 참고하세요. X에는 표의 입력 특성 두 열, y에는 실제 구매 여부 한 열을 담습니다. 고객 C의 첫 번째 3은 방문 횟수이고 두 번째 3은 장바구니 추가 횟수입니다. 각각 어떤 가중치를 곱해야 하는지 확인하세요.

#### 해석 질문

**Q1.** B의 입력 벡터는 무엇인가요? 그 안의 두 숫자는 각각 무엇을 뜻하나요?

**Q3.** C의 점수는 얼마인가요? ‘첫 곱셈 → 둘째 곱셈 → 편향을 더한 합’ 순서로 계산하고, 이 값이 구매 확률인지 답하세요.

**Q4.** 가중치와 편향은 무엇이며, 이번 코드에서 그 값들을 학습으로 바꾸었나요?

- x_b, X, y, w, b와 Shape 출력

- Q1~Q4 답변

```python
# 문제 3-1. 이번 문제의 데이터로 배열을 새로 만듭니다.
x_b = np.array([6, 1])
X = np.array([[4, 2], [6, 1], [3, 3]])
y = np.array([1, 0, 1])
w = np.array([2, 3])
b = 1

print("x_b:", x_b)
print("X:\n", X, "\nshape:", X.shape)
print("y:", y, "/ shape:", y.shape)
print("w:", w, "/ shape:", w.shape)
print("b:", b)

x_c = np.array([3, 3])
score_c = 2 * 3 + 3 * 3 + 1
print("x_c:", x_c)
print("score_c = 2 * 3 + 3 * 3 + 1 =", score_c)
```

**Q2. X, y, w의 Shape는 각각 무엇인가요? 각 배열에 무엇이 몇 개 들어 있는지 설명하세요.**

### 문제 3-2. 고객 B 한 벡터의 노름과 정규화 복습

#### 문제 설명

교안 **연습문제 4**의 고객 B를 다시 사용합니다. 필수 2의 계산 순서를 익히는 기초 과제입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| B | 6 | 8 |

`x_b=[6,8]`이며 첫 원소는 방문 횟수, 둘째 원소는 장바구니 추가 횟수입니다.

- L1 Norm: 각 원소의 절댓값을 더합니다.
- L2 Norm: 각 원소를 제곱하고, 더한 뒤, 제곱근을 구합니다.
- L2 정규화: 각 원소를 그 벡터의 L2 Norm으로 나눕니다.

#### 요구사항

1. `x_b`를 NumPy 배열로 만들고 출력하세요.
2. `np.linalg.norm(x_b, ord=1)`과 `np.linalg.norm(x_b, ord=2)`로 L1·L2 Norm을 구하여 `b_l1`, `b_l2`에 저장하고 출력하세요.
3. `b_normalized = x_b / b_l2`로 정규화하고 출력하세요.
4. 정규화한 벡터의 L2 Norm을 계산하여 출력하세요. 각 단계의 손계산도 답변란에 적으세요.

#### 작성 도움말

순서는 **원래 벡터 → L1·L2 Norm → L2 Norm으로 나누기 → 새 길이 확인**입니다. 처음부터 공식을 떠올리지 않아도 됩니다. 위 설명과 필수 2-1·2-2를 참고하여 숫자만 대입하세요.

#### 해석 질문

**Q1.** B의 L1 Norm은 얼마인가요? 두 절댓값과 그 합을 차례로 적으세요.

**Q2.** B의 L2 Norm은 얼마인가요? 제곱, 합, 제곱근 순서로 계산하세요.

**Q3.** 정규화할 때 무엇으로 나누나요? 두 원소를 각각 나누어 결과를 적으세요.

**Q4.** 정규화한 벡터의 L2 Norm은 얼마인가요? 길이가 1이라는 것이 두 원소의 합이 1이라는 뜻인가요?

#### 제출 결과

- 원래 벡터와 L1·L2 Norm 출력
- 정규화 벡터와 정규화 후 L2 Norm 출력
- 손계산 과정과 Q1~Q4 답변

```python
# 문제 3-2. 원래 크기를 구한 뒤 L2 정규화를 적용합니다.
x_b = np.array([6, 8])
b_l1 = np.linalg.norm(x_b, ord=1)
b_l2 = np.linalg.norm(x_b, ord=2)
b_normalized = x_b / b_l2
normalized_l2 = np.linalg.norm(b_normalized, ord=2)

print("x_b:", x_b)
print("b_l1:", b_l1)
print("b_l2:", b_l2)
print("b_normalized:", b_normalized)
print("정규화 후 L2 Norm:", normalized_l2)
```

#### 풀이 및 답변

**수식 및 계산 과정**

- $\|x_B\|_1=|6|+|8|=6+8=14$.
- $\|x_B\|_2=\sqrt{6^2+8^2}=\sqrt{36+64}=\sqrt{100}=10$.
- L2 정규화 결과는 `[6 / 10, 8 / 10] = [0.6, 0.8]`입니다.
- 정규화 후 길이는 $\sqrt{0.6^2+0.8^2}=\sqrt{0.36+0.64}=1$입니다.

**Q1. B의 L1 Norm은 얼마인가요? 두 절댓값과 그 합을 차례로 적으세요.**

- **답변:** L1 Norm은 14입니다. $|6|=6$, $|8|=8$이므로 합은 $6+8=14$입니다.

**Q2. B의 L2 Norm은 얼마인가요? 제곱, 합, 제곱근 순서로 계산하세요.**

- **답변:** L2 Norm은 10입니다. $6^2=36$, $8^2=64$, 제곱합은 100이며 제곱근은 $\sqrt{100}=10$입니다.

**Q3. 정규화할 때 무엇으로 나누나요? 두 원소를 각각 나누어 결과를 적으세요.**

- **답변:** 원래 벡터 자신의 L2 Norm인 10으로 나눕니다. 첫 원소는 $6/10=0.6$, 둘째 원소는 $8/10=0.8$이므로 결과는 `[0.6, 0.8]`입니다.

**Q4. 정규화한 벡터의 L2 Norm은 얼마인가요? 길이가 1이라는 것이 두 원소의 합이 1이라는 뜻인가요?**

- **답변:** L2 Norm은 1입니다. 길이가 1이라는 것은 제곱합의 제곱근이 1이라는 뜻입니다. 원소의 합은 $0.6+0.8=1.4$이므로, 원소의 합이 1이라는 뜻은 아닙니다.
