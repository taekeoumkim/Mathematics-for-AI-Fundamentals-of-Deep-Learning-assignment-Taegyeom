### 문제 1-2. Sigmoid 확률과 Threshold로 구매 여부 결정하기

#### 문제 설명

문제 1-1의 고객별 분류 점수를 사용합니다. 계산의 출발점을 분명히 하기 위해 점수를 다시 배열로 만드세요.

| 고객 | 점수 z |
|---|---:|
| A | 0.8 |
| B | −1.4 |
| C | 2.2 |

교안의 공식 `Sigmoid(z) = 1 / (1 + e^(−z))`로 구매 확률 p를 계산합니다. e는 약 2.718인 자연상수이며, `np.exp(-z)`가 e^(−z)를 계산합니다. 미구매 확률은 `1 − p`입니다.

결정 규칙은 **p가 임계값 이상이면 구매(1), 작으면 미구매(0)**입니다. 임계값은 0.5와 0.7을 각각 적용합니다. 추가로 점수가 정확히 0인 경우도 확인합니다.

#### 수식과 기호 읽기

`1 / (1 + e^(−z))`는 **z의 부호를 바꾼다 → e의 그 제곱값을 구한다 → 1을 더한다 → 그 전체로 1을 나눈다**는 순서입니다. 분수의 위 숫자는 분자, 아래 숫자는 분모입니다. 여기서는 분자가 1, 분모가 `1+e^(−z)`입니다.

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

함수는 위 코드를 그대로 사용해도 됩니다. z가 −1.4이면 −z는 +1.4입니다. 음수 앞에 다시 마이너스를 붙이면 양수가 됩니다.

````python
# 문제 1-2. 점수를 확률로 바꾼 뒤 임계값과 비교합니다.
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.array([0.8, -1.4, 2.2])
p = sigmoid(z)
non_purchase_prob = 1 - p
# 반올림은 출력에만 사용합니다.
pred_05 = (p >= 0.5).astype(int)
pred_07 = (p >= 0.7).astype(int)
p_zero = sigmoid(0)
pred_zero = int(p_zero >= 0.5)

print("구매 확률:", np.round(p, 3))
print("미구매 확률:", np.round(non_purchase_prob, 3))
print("임계값 0.5의 예측:", pred_05)
print("임계값 0.7의 예측:", pred_07)
print(f"Sigmoid(0): {p_zero:.3f}")
print("점수 0, 임계값 0.5의 예측 클래스:", pred_zero)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- $p_A=1/(1+e^{-0.8})\approx0.689974$, $p_B=1/(1+e^{1.4})\approx0.197816$, $p_C=1/(1+e^{-2.2})\approx0.900250$입니다.
- 구매 확률을 세 자리로 표시하면 `[0.690, 0.198, 0.900]`, 미구매 확률은 `[0.310, 0.802, 0.100]`입니다.
- 반올림 전 확률에 대해 임계값 0.5의 예측은 `[1, 0, 1]`, 임계값 0.7의 예측은 `[0, 0, 1]`입니다.
- $\mathrm{Sigmoid}(0)=1/(1+e^0)=1/(1+1)=0.5$입니다. `0.5 >= 0.5`가 참이므로 예측 클래스는 1입니다.

- **답변:** 구매 확률은 약 0.690(69.0%), 미구매 확률은 약 0.310(31.0%)입니다. 점수 0.8 자체와 Sigmoid를 거친 확률은 구분해야 합니다.

- **답변:** B의 구매 확률은 약 0.198, C는 약 0.900입니다. Sigmoid는 점수가 클수록 더 큰 확률을 출력하므로 B < A < C의 순서가 유지됩니다.

- **답변:** 0.5에서는 A·B·C가 `[1, 0, 1]`, 0.7에서는 `[0, 0, 1]`입니다. A의 확률 약 0.689974가 0.5 이상이면서 0.7보다 작아 A만 1에서 0으로 바뀝니다.

- **답변:** 0의 지수값은 $e^0=1$이므로 $1/(1+1)=0.5$입니다. 규칙이 임계값 이상을 1로 정하는 `>=`이므로, 정확히 0.5인 경우도 구매(1)로 분류합니다.

교안 2·3절의 예제입니다. 한 상품을 의류·식품·가전 중 하나로 분류합니다.

| 클래스 | 의류 | 식품 | 가전 |
|---|---:|---:|---:|
| 인덱스 | 0 | 1 | 2 |
| 점수 z | 2 | 1 | 0 |

Softmax는 **각 점수의 지수값을 구하고 → 그 지수값들을 더하고 → 각 지수값을 합계로 나누는 계산**입니다. 수식에서 Σ(시그마)는 모두 더하라는 뜻입니다.

`Softmax(zᵢ) = e^(zᵢ) / Σ e^(zⱼ)`

i는 지금 확률을 구하려는 클래스의 위치, j는 합계에 넣는 모든 클래스의 위치입니다. 아래 코드는 교안처럼 가장 큰 점수를 먼저 빼고 계산합니다. 모든 점수에서 같은 값을 빼도 Softmax 확률은 유지됩니다.

```python
def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)
```

함수는 그대로 사용하세요. 이번 문제에서는 이 성질을 증명하지 않고 계산 순서를 따라갑니다.

#### 요구사항

2. 위 함수를 작성하고 `p = softmax(z)`를 계산하세요.
3. `np.round(p, 3)`과 `np.sum(p)`로 확률과 확률 합을 출력하세요.
4. `index = np.argmax(p)`로 인덱스를 구하고 index와 `classes[index]`를 출력하세요.

#### 작성 도움말

최댓값 2를 각 점수에서 빼면 [0,−1,−2]입니다. 지수값은 약 [1,0.3679,0.1353]이고 합은 약 1.5032입니다. 각 지수값을 이 합으로 나누세요. `np.argmax(p)`는 가장 큰 확률이 있는 위치를 반환합니다.

**Q3.** Argmax 결과와 예측 클래스 이름은 무엇인가요? Argmax가 확률값을 반환하나요?

**Q4.** 이 문제에서 Softmax를 사용하는 이유는 무엇인가요?

#### 제출 결과

- Softmax 함수와 확률·확률 합
- 예측 인덱스·클래스 이름 및 Q1~Q4 답변

````python
# 문제 2-1. 한 상품의 세 점수에 Softmax를 적용합니다.
def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)

z = np.array([2, 1, 0])
classes = np.array(["의류", "식품", "가전"])
shifted = z - np.max(z)
exp_z = np.exp(shifted)
p = softmax(z)
index = np.argmax(p)

print("최댓값을 뺀 점수:", shifted)
print("지수값:", np.round(exp_z, 4))
print("클래스별 확률:", np.round(p, 3))
print("확률 합:", np.sum(p))
print("예측 인덱스:", index)
print("예측 클래스:", classes[index])
````

#### 풀이 및 답변

**수식 및 계산 과정**

- 최댓값 2를 빼면 `[0, -1, -2]`입니다. 지수값은 $[1,e^{-1},e^{-2}]\approx[1,0.367879,0.135335]$이고 합은 약 1.503215입니다.
- 각 지수값을 합으로 나누면 `[0.665241, 0.244728, 0.090031]`입니다. 세 자리로 표시한 확률은 `[0.665, 0.245, 0.090]`이고, 원래 확률의 합은 수치 오차 범위에서 1입니다.
- 가장 큰 확률은 첫 위치이므로 `index = 0`, `classes[0] = "의류"`입니다.

**Q2. 세 클래스의 확률은 각각 얼마이며, 확률 합은 얼마인가요?**

### 문제 2-2. 여러 고객의 확률과 클래스를 행별로 구하기

#### 문제 설명

교안 2·3절의 고객별 점수 예제입니다. 가로 한 줄인 행은 고객, 세로 한 줄인 열은 클래스입니다.

```python
Z = np.array([[2, 1, 0], [0, 2, 1], [1, 0, 3]])
classes = np.array(["의류", "식품", "가전"])
```

첫 행은 A, 둘째 행은 B, 셋째 행은 C입니다. 클래스 순서는 의류·식품·가전입니다. 한 고객의 세 점수끼리 Softmax를 계산합니다.

아래 교안의 함수를 **그대로 사용**하세요. 여러 행을 계산하는 데 필요한 옵션은 해설에서 다시 풀이합니다.

```python
def softmax(z):
    shifted = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(shifted)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
```

`axis=1`은 각 행 안의 세 클래스 값을 대상으로 계산하라는 뜻입니다. `keepdims=True`는 고객별로 구한 값이 자신의 행에 적용되도록 결과의 모양을 유지합니다.

3. `Z.shape`, `p.shape`, `np.sum(p, axis=1)`을 출력하세요.
4. `predicted_indices = np.argmax(p, axis=1)`를 계산하고, 인덱스와 `classes[predicted_indices]`를 출력하세요.

#### 작성 도움말

한 행만 먼저 읽어 보세요. 첫 행 [2,1,0]은 앞 문제와 같으므로 같은 확률이 나옵니다. 고객별로 가장 큰 확률이 의류·식품·가전 중 어느 열에 있는지 찾습니다. 함수 안의 axis와 keepdims는 제시된 코드대로 사용하면 됩니다.

#### 해석 질문

**Q1.** Z와 p의 Shape는 무엇이며, 행과 열은 각각 무엇인가요?

**Q3.** 고객 A·B·C의 예측 인덱스와 클래스 이름은 각각 무엇인가요?

**Q4.** 이 문제에서 axis=1을 지정하는 이유는 무엇인가요?

#### 제출 결과

- 고객별 확률, Z·p의 Shape와 고객별 확률 합
- 고객별 예측 인덱스·이름 및 Q1~Q4 답변

````python
# 문제 2-2. 고객마다 자신의 세 클래스 점수로 계산합니다.
def softmax(z):
    shifted = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(shifted)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)

Z = np.array([[2, 1, 0], [0, 2, 1], [1, 0, 3]])
classes = np.array(["의류", "식품", "가전"])
p = softmax(Z)
row_sums = np.sum(p, axis=1)
predicted_indices = np.argmax(p, axis=1)
predicted_classes = classes[predicted_indices]

print("고객별 클래스 확률:\n", np.round(p, 3))
print("Z.shape:", Z.shape)
print("p.shape:", p.shape)
print("고객별 확률 합:", row_sums)
print("고객별 예측 인덱스:", predicted_indices)
print("고객별 예측 클래스:", predicted_classes)
````

#### 풀이 및 답변

**수식 및 계산 과정**

- B: `[0, 2, 1] → [-2, 0, -1] → [0.090031, 0.665241, 0.244728]`.
- C: `[1, 0, 3] → [-2, -3, 0] → [0.114195, 0.042010, 0.843795]`. 각 행에서 지수값을 구한 뒤 그 행의 지수값 합으로 나눴습니다.
- 행별 확률 합은 `[1, 1, 1]`, 예측 인덱스는 `[0, 1, 2]`, 클래스는 `[의류, 식품, 가전]`입니다.

**Q1. Z와 p의 Shape는 무엇이며, 행과 열은 각각 무엇인가요?**

- **답변:** Z와 p의 Shape는 모두 `(3, 3)`입니다. 각 행은 고객 A·B·C이고, 각 열은 의류·식품·가전 클래스입니다.

**Q2. 세 고객의 확률 분포와 각 고객의 확률 합은 무엇인가요?**

- **답변:** A는 약 `[0.665, 0.245, 0.090]`, B는 `[0.090, 0.665, 0.245]`, C는 `[0.114, 0.042, 0.844]`입니다. 반올림 전 확률의 합은 고객마다 1입니다.

**Q3. 고객 A·B·C의 예측 인덱스와 클래스 이름은 각각 무엇인가요?**

- **답변:** A는 인덱스 0인 의류, B는 인덱스 1인 식품, C는 인덱스 2인 가전으로 예측합니다.

**Q4. 이 문제에서 axis=1을 지정하는 이유는 무엇인가요?**

- **답변:** `axis=1`은 각 행 안에서 클래스 방향으로 계산한다는 뜻입니다. Softmax에서는 고객별 최댓값과 지수값 합을, Argmax에서는 고객별로 가장 큰 확률의 클래스 위치를 구합니다. `keepdims=True`는 고객별 값이 자신의 행에 적용되도록 모양을 유지합니다.

### 문제 3-2. 한 상품의 Softmax와 Argmax 복습하기

|---|---:|---:|---:|

이 한 상품은 세 종류 중 하나로 분류합니다. 교안 3절의 Argmax를 사용해 가장 높은 확률의 클래스도 확인하세요.

#### 요구사항

1. 문제 2-1의 한 벡터용 Softmax 함수를 그대로 작성하세요.

#### 작성 도움말

한 벡터용 함수에는 axis=1을 쓰지 않습니다. 가장 큰 점수 2를 빼면 [−1,0,−1]이고, 지수값은 약 [0.3679,1,0.3679]입니다. 이 세 값을 더한 뒤 각각을 같은 합으로 나누세요.

#### 해석 질문

**Q1.** 최댓값을 뺀 배열, 지수값, 지수값의 합은 각각 무엇인가요?

**Q4.** 의류와 가전의 확률이 같은 이유는 무엇인가요?

#### 제출 결과

- 한 벡터용 Softmax 함수, 확률·확률 합
- 예측 인덱스·클래스 이름과 Q1~Q4 답변

````python
# 문제 3-2. 한 벡터용 Softmax 함수를 다시 정의합니다.
def softmax(z):
    exp_z = np.exp(z - np.max(z))
    return exp_z / np.sum(exp_z)

z = np.array([1, 2, 1])
classes = np.array(["의류", "식품", "가전"])
shifted = z - np.max(z)
exp_z = np.exp(shifted)
exp_sum = np.sum(exp_z)
p = softmax(z)
index = np.argmax(p)

print("최댓값을 뺀 점수:", shifted)
print("지수값:", np.round(exp_z, 4))
print(f"지수값의 합: {exp_sum:.6f}")
print("클래스별 확률:", np.round(p, 3))
print("확률 합:", np.sum(p))
print("예측 인덱스:", index)
print("예측 클래스:", classes[index])
````

**수식 및 계산 과정**

- 최댓값 2를 빼면 `[-1, 0, -1]`, 지수값은 약 `[0.367879, 1, 0.367879]`이고 합은 약 1.735759입니다.
- 확률은 약 `[0.211942, 0.576117, 0.211942]`이며, 세 자리로 표시하면 `[0.212, 0.576, 0.212]`입니다.
- 확률 합은 수치 오차 범위에서 1이고, 최댓값의 인덱스는 1이므로 최종 예측은 식품입니다.

**Q2. 의류·식품·가전의 확률과 확률 합은 각각 얼마인가요?**

- **답변:** 의류는 약 0.212, 식품은 약 0.576, 가전은 약 0.212이며, 반올림 전 확률의 합은 1입니다.

**Q3. Argmax 결과와 최종 예측 클래스는 무엇인가요? 인덱스 1은 확률 100%라는 뜻인가요?**

- **답변:** Argmax는 1을 반환하고 최종 예측은 식품입니다. 인덱스 1은 두 번째 위치를 뜻하며 확률 100%라는 의미가 아닙니다. 식품에 부여된 확률은 약 57.6%입니다.

**2. Sigmoid(0)은 얼마이며, 임계값 0.5에서는 어떤 클래스로 정하나요?**

- **답변:** Sigmoid(0)은 $1/(1+e^0)=0.5$입니다. 이 실습의 규칙은 `p >= 0.5`이면 1이므로 구매(1)로 정합니다.

**3. 구매 확률이 p일 때 미구매 확률은 무엇인가요?**

- **답변:** 미구매 확률은 $1-p$입니다. 구매와 미구매 두 가지 결과의 확률을 더하면 1입니다.

**4. Softmax는 무엇을 더한 합으로 각 지수값을 나누나요?**

- **답변:** 같은 데이터에 대한 모든 클래스의 지수값을 더한 합으로 나눕니다. 여러 고객이면 고객마다 자신의 행 안에서 합을 구합니다.

**5. 여러 고객의 확률에서 Argmax의 axis=1은 무엇을 뜻하나요?**

**6. 임계값을 높이거나 가장 높은 확률을 고르면 반드시 더 정확한 예측인가요?**

- **답변:** 정확도가 반드시 높아지지는 않습니다. 임계값이나 Argmax는 확률을 클래스로 바꾸는 결정 규칙이며, 실제로 맞았는지는 정답과 비교해야 알 수 있습니다.
